Insert / Delete/ Append 

```python
import numpy as np


A = np.array([
    [1,2,3],
    [10,20,30],
    [100,200,300]
])
```

```python
#insert 
#axis => 0 > row
# axis => 1 > column 
np.insert(A,0,[1000,2000,3000], axis = 0)
```

```python
np.insert(A,0, [101,102,103], axis = 1)
```

```python
#delete row
np.delete(A,0,axis = 0)
```

```python
np.delete(A,1,axis = 1)
```

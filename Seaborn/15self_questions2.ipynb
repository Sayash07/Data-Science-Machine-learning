Automatically identify all categorical columns and create a countplot for each categorical variable in one figure using enumerate().


```python
from matplotlib import pyplot as plt
import seaborn as sns


df = sns.load_dataset("tips")
```

```python
categorcial_data = df.select_dtypes(include ="category")

for i,value in enumerate(categorcial_data):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[value])

plt.tight_layout()
plt.show()
```

Create a boxplot of total_bill grouped by day.

```python
sns.boxplot(y=df["total_bill"],hue= df["day"])
plt.tight_layout()
plt.show()
```

Create a barplot showing the average tip for smokers and non-smokers.


```python
sns.barplot(y=df["tip"],x=df["smoker"])
plt.show()
```

Create a scatterplot between total_bill and tip, using smoker as the hue.


```python
sns.scatterplot(x=df["total_bill"],y=df["tip"],hue=df["smoker"])
plt.show()
```

Create a boxplot showing tip across different days, with separate boxes for smokers and non-smokers.


```python
sns.boxplot(y=df["tip"],x=df["day"],hue=df["smoker"])
plt.show()
```

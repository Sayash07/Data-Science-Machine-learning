```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.DataFrame({
    "Area": [500, 700, 900, 1200, 1500, 2000, 3000, 5000],
    "Bedrooms": [1, 2, 2, 3, 3, 4, 5, 6],
    "Price": [30000, 40000, 50000, 70000, 90000, 130000, 250000, 600000]
})

print(df)
```

```python
from sklearn.preprocessing import FunctionTransformer


feature =["Bedrooms"]
square_transform = FunctionTransformer(lambda x: x**2)

square_feature = square_transform.fit_transform(df[feature])
df[feature] = square_feature
df
```

Find the average tip for each combination of day and time, then visualize the result.


```python
import seaborn as sns
from matplotlib import pyplot as plt
df= sns.load_dataset("tips")
df
```

```python


sns.barplot( x =df["day"],y=df["tip"],hue = df["time"])
plt.tight_layout()
plt.show()
```

Calculate the average tip percentage for each day using:

```python
#tip_percentage = (tip / total_bill) * 100
tip_percentage=(df["tip"]/df["total_bill"])*100

sns.barplot(x=df['day'],y=tip_percentage)
plt.show()
```

Determine whether tip appears to increase as total_bill increases. Create an appropriate visualization and calculate the correlation between them.

```python
num_data = df.select_dtypes(include = "float64")

correlation = num_data.corr()
correlation

sns.heatmap(correlation, annot = True)
plt.show()
```

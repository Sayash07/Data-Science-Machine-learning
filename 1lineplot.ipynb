```python
day=["sun","mon", "tue","wed","thu","fri","sat"]
share_price=[200,205,210,210,200,225,205]

```

```python
#lineplot
from matplotlib import pyplot as plt

plt.plot(day,share_price)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NTC")
#Changing xtick:
new_list = ["Sunday","Monday","Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
plt.xticks(day,new_list)
plt.xticks(rotation = 45)
plt.show()
```

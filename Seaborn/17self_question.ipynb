Create an exploratory data analysis (EDA) dashboard for the tips dataset. It should contain at least 6 subplots and answer questions about:
Which day has the most customers?
How are total bills distributed?
How are tips distributed?
Do larger bills tend to have larger tips?
Do smokers and non-smokers have different bill distributions?
Is there a difference between lunch and dinner?

```python
from matplotlib import pyplot as plt
import seaborn as sns
df = sns.load_dataset("tips")
df
```

```python
plt.figure(figsize=(10,5))
# 1. Which day has the most customers?
plt.subplot(3,2,1)
sns.countplot(x =df["day"])
# 2. Distribution of total_bill
plt.subplot(3,2,2)
sns.histplot(x=df["total_bill"])
# 3. Distribution of tip
plt.subplot(3,2,3)
sns.histplot(x=df["tip"])
# 4. Relationship between total_bill and tip
plt.subplot(3,2,4)
sns.scatterplot(x=df["total_bill"], y= df["tip"])
# 5. Bill distribution for smokers vs non-smokers
plt.subplot(3,2,5)
sns.barplot(y=df["total_bill"], x=df["smoker"])
# 6. Difference between lunch and dinner
plt.subplot(3,2,6)
sns.boxplot(y=df["time"],x=df["total_bill"])


plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
from matplotlib import pyplot as plt
df = sns.load_dataset("tips")
print(df.head())
df.info()
```

```python
# make a countplot for all categorical variable
#make histplot for all numeric variable(single variable)



plt.figure(figsize=(10, 5))
categorical_value = df.select_dtypes(include= "category").columns

plt.subplot(2,2,1)
sns.countplot(x= df[categorical_value[0]])
plt.subplot(2,2,2)
sns.countplot(x= df[categorical_value[1]])
plt.subplot(2,2,3)
sns.countplot(x= df[categorical_value[2]])
plt.subplot(2,2,4)
sns.countplot(x= df[categorical_value[3]])


plt.show()
plt.tight_layout()
```

```python
plt.figure(figsize=(10, 5))
num_value = df.select_dtypes(include= ['int64','float64']).columns

plt.figure(figsize=(10, 5))
plt.subplot(2,2,1)
sns.histplot(x=df[num_value[0]])
plt.subplot(2,2,2)
sns.histplot(x=df[num_value[1]])
plt.subplot(2,2,3)
sns.histplot(x=df[num_value[2]])


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
categorical_value= df.select_dtypes("category").columns

for i,value in enumerate(categorical_value):
    
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[value])
    
    
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize= (10,5))
num_catergory = df.select_dtypes(include= ("int64","float64")).columns
print(num_catergory)

for i,value in enumerate(num_catergory):
    plt.subplot(2,2,i+1)
    sns.histplot(x=df[num_catergory[i]])

plt.tight_layout()
plt.show()
```

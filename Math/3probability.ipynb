P(A)= Number of favorable outcomes/Total number of possible outcomes
Lies between 0 and 1

>> Tossing a coin
P(Head) = 1/2

>> Rolling a dice
1 2 3 4 5 6
P(4)= 1/6

>> Drawing Card
P(Red)= 26/52
P(Black)= 26/52

Independent and Dependent event :

>> Two events are independent if one event does not affect the probability of the other.
Probability of Head and again Head:
P(H n H) = P(H) x P(H) 
= (1/2) * (1/2)


>> Two events are dependent if one event affects the probability of the other.
Probability of a card is red while drawing from a deck of cards
1st event P(Red) = 26/52  | 2nd event P(Red) = 25/51
P(Red n Red) = 26/52 * 25/51

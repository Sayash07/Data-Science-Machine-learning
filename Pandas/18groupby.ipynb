```python
import pandas as pd
data = {
    "Name": ['nitan', 'sita', 'gita',"ram"],
    "Gender": ["male","female","female","male"],
    "Salary": [10000, 20000, 30000,10000]
}
df=pd.DataFrame(data)
df
```

```python
df.groupby("Gender")["Salary"].mean()
```

```python
df.groupby("Gender")["Salary"].transform("mean")
```

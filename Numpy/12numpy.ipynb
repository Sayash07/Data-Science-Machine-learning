Broadcasting >> expands to similar 
1 >> [1,1,1]

(2,3) & (2,3) => ok
(2,3) & (2,4) => not ok


```python
import numpy as np


ar1 = np.array([1,2,3])

print(ar1 +1)
```

```python
ar2 = np.array([1,2],[3,4])
ar3 = np.array([1,5])

ar2 + ar3

""" 
1 2     1 5     2 7
3 4     1 5     4 9

 """


```

Rules of Broadcasting:
2,3 & 2,3 => ok
2,3 & 2,4 => not ok
2,3 & 1,1 =>  ok
2,3 & 1,3 =>  ok
2,3 & 3,3 => not ok
1,2 & 3,1 => ok
3,3 & 3,2 => not ok
3,1 & 2 => 3,1 & 1,2 => ok
either corresponding elements must be equal or atleast 1 of them must be 1

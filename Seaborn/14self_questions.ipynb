```python
from matplotlib import pyplot as plt
import seaborn as sns

df = sns.load_dataset("tips")
df
```

Create a countplot showing how many customers visited on each day.


```python
sns.countplot(x='day', data=df)
plt.show()


```

Create a histogram showing the distribution of total_bill.


```python
sns.histplot(data = df, x=df["total_bill"])
plt.show()
```

Create a histogram showing the distribution of tip using 15 bins.

```python
sns.histplot(data = df, x=df["tip"], bins = [0,5,10,15,20,25])
plt.show()
```

Create a countplot showing the number of customers who are smokers vs non-smokers.


```python
sns.countplot(data= df, x=df["smoker"])
plt.show()
```

Create a boxplot for total_bill and identify potential outliers visually.


```python
sns.boxplot(data= df, x=df["total_bill"])
plt.show()
```

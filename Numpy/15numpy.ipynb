```python
import numpy as np


students = np.array([
    [1,56,99],
    [2,52,80],
    [3,90,48],
    [4,78,81],
])
```

```python
#Filer those row whose maths marks are greater than 60

students[students[:,1] >= 60]
```

```python
#filter those row whose both marks >= 80:

students[(students[:,1] >= 60) & (students[:,2] >= 60)]
```

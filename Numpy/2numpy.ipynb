```python
#Dimension
#1D array => vector
#2D array => matrix
```

```python
import numpy as np

l1 = np.array([10, 11, 12]) # 1d array
l2 = np.array([[1,2,3],[4,5,6]]) #2D array(Array with array elements)
```

```python
# Shape
print(l1.shape)
print(l2.shape) 
```

```python
# (2,) => ([1,2])
#(3,4) => ([1,2,3,4],[2,3,4,5],[3,4,5,6])
#(3,3) => ([1,2,3],[2,3,4],[3,4,5])

```

```python
#dtype (It gives data type of most inner element)
l1 = np.array([1,2,3])
print(l1.dtype)
l2 = np.array([[1,2,3],[4,5,6]])
print(l2.dtype)
```

#Creating numpy array
#zeros

```python
#(20,)
import numpy as np

l1 = np.zeros((20,))
print(l1)
l2 = np.ones((20,))
print(l2)
l3 = np.random.randint(1,20,(20,))
print(l3)
# (20,0) => 20 // (20,2) => 20,2
l4 = np.random.rand(20) #rand => from 0 to 1
print(l4)

```

```python
# (3,4)
l5 = np.random.randint(1,50,(3,4))
print(l5)
```

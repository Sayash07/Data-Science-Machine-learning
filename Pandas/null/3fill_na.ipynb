```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari", "Sita", "Gita"],
    "Age": [20, np.nan, 22, None, 21],
    "Math": [80, 75, np.nan, 90, None],
    "Science": [85, np.nan, 78, 92, 88],
    "City":["Kathmandu","Pokhara","Lalitpur",None,"Lalitpur"]
})

print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Math"]= df["Math"].fillna(df["Math"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())
df["City"]=df["City"].fillna(df["City"].mode()[0]) # [0] >> gives first mode if there is tie
df
```

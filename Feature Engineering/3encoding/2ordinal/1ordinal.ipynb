```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35, 40, 27, 45],
    "Experience": [1, 2, 4, 5, 8, 12, 3, 15],
    "Education": [
        "Bachelor", "Bachelor", "Master", "Master",
        "Bachelor", "Master", "Bachelor", "Master"
    ],
    "Performance Level": [
        "Poor", "Average", "Good", "Excellent",
        "Good", "Excellent", "Average", "Excellent"
    ]
})

print(df)
```

```python
from sklearn.preprocessing import OrdinalEncoder


feature = ['Education', 'Performance Level']
order = [
    ['Bachelor', 'Master'],
    ['Poor', 'Average', 'Good', 'Excellent']
]

encoder = OrdinalEncoder(categories= order)
encoded_feature = encoder.fit_transform(df[feature])
encoded_feature
```

```python
df[encoder.get_feature_names_out()]= encoded_feature
df
```

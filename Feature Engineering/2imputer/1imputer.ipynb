```python
#before diving first learn train_test_split.png image  

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman"
    ],

    # Categorical variables
    "Gender": [
        "Male", "Male", np.nan, "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],  # 10% missing

    "City": [
        "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
        "Pokhara", "Lalitpur", "Kathmandu", "Pokhara", np.nan
    ],  # 10% missing

    "Department": [
        "IT", np.nan, np.nan, "Finance", "IT",
        np.nan, "HR", np.nan, "IT", "Finance"
    ],  # 40% missing

    # Numerical variables
    "Age": [
        20, 22, np.nan, 21, 23,
        24, 25, 26, 22, 21
    ],  # 10% missing

    "Math": [
        80, 75, 90, np.nan, 85,
        78, 88, 92, 70, 95
    ],  # 10% missing

    "Salary": [
        30000, np.nan, np.nan, 40000, np.nan,
        35000, np.nan, 50000, np.nan, 38000
    ]  # 50% missing 
})

df
```

```python
df.isna().sum()
```

```python
df_numerical = df.select_dtypes(include = ['int64','float64'])
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy = "mean")
df_numerical_imputed = imputer.fit_transform(df_numerical)
df_numerical_imputed.round(1)
```

```python
df[imputer.get_feature_names_out()] = df_numerical_imputed #It gives columns names
df
```

```python
test_df = pd.DataFrame({
    "Name": [
        "Aashish", "Kiran"
    ],

    "Gender": [
        "Male", np.nan
    ],

    "City": [
        "Pokhara", "Kathmandu"
    ],

    "Department": [
        np.nan, "Finance"
    ],

    "Age": [
        23, np.nan
    ],

    "Math": [
        88, 76
    ],

    "Salary": [
        np.nan, 42000
    ]
})
test_df
```

```python
#get mean from train data and fill test data 

df_numerical_test = test_df.select_dtypes(include=['int64','float64'])

df_numerical_test_imputed = imputer.transform(df_numerical_test)

test_df[imputer.get_feature_names_out()] = df_numerical_test_imputed
test_df
```

```python
#fill categorical value:
df_categorical = df.select_dtypes(include= "str")
imputer1 = SimpleImputer(strategy='most_frequent')
df_categorical_imputed = imputer1.fit_transform(df_categorical)
df[imputer1.get_feature_names_out()]= df_categorical_imputed
df
```

```python
#test_df:
test_df[imputer1.get_feature_names_out()] = imputer1.transform(test_df[imputer1.get_feature_names_out()])
test_df
```

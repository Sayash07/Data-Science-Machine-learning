```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Nabin", "Asha", "Rohan", "Mina"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Math": [85.2134, 90.67812, 70, 95, 60, 88, 75, 92],
    "Science": [80, 85, 75, 90, 65, 95, 70, 89],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Lalitpur",
             "Bhaktapur", "Pokhara", "Kathmandu", "Lalitpur"]
}

df = pd.DataFrame(data)

df
```

```python
df[df["Math"] >= 80]
```

```python
#Filter those students whose both marks >= 80 and is male
df[(df["Math"]>=80)&(df["Science"]>=80)&(df["Gender"]=="M")]
```

```python
#Filter those whose age is 20 or 22
df[(df["Age"] == 20) | (df["Age"] == 22)]
```

```python
#Filter whose city is not kathmandu
df[~(df["City"]=="Kathmandu")]
```

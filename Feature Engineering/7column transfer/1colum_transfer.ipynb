```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman"
    ],

    # Categorical variables
    "Gender": [
        "Male", "Male", np.nan, "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],  # 10% missing

    "City": [
        "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
        "Pokhara", "Lalitpur", "Kathmandu", "Pokhara", np.nan
    ],  # 10% missing

    "Department": [
        "IT", np.nan, np.nan, "Finance", "IT",
        np.nan, "HR", np.nan, "IT", "Finance"
    ],  # 40% missing

    # Numerical variables
    "Age": [
        20, 22, np.nan, 21, 23,
        24, 25, 26, 22, 21
    ],  # 10% missing

    "Math": [
        80, 75, 90, np.nan, 85,
        78, 88, 92, 70, 95
    ],  # 10% missing

    "Salary": [
        30000, 33000, 33000, 40000, 34000,
        35000, 33000, 50000, 35000, 38000
    ]  # 50% missing 
})
df.info()
```

```python
numerical_columns = df.select_dtypes(include = ['int64','float64']).columns
categorical_columns = df.select_dtypes(include= ['str']).columns

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer


preprocessor = ColumnTransformer(
    [
        ("num", SimpleImputer(strategy='mean'), numerical_columns),

        ("cat", SimpleImputer(strategy='most_frequent'), categorical_columns), 
    ],         
    verbose_feature_names_out= False #It doesn't change the feature names


)
```

```python
value = preprocessor.fit_transform(df)
df[preprocessor.get_feature_names_out()]= value
df
```

```python
import pandas as pd


df = pd.DataFrame({
    "age": [20, 22, 25, 28, 30, 32, 35, 38, 40, 45],
    "income": [20000, 25000, 30000, 35000, 40000,
               45000, 50000, 55000, 60000, 70000],
    "purchased": ["No", "No", "No", "Yes", "No",
                  "Yes", "Yes", "Yes", "Yes", "Yes"]
})
df 
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
sns.scatterplot(x=df["age"],y=df["income"],hue = df["purchased"]) # Hue is used to seperate points
plt.show()
```

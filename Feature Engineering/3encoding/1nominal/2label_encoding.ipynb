```python
import pandas as pd

df = pd.DataFrame({
    "Study Hours": [2, 3, 4, 5, 6, 7, 1, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 50, 90],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Result": ["Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Fail", "Pass"]
})

print(df)
```

```python
feature = ['Result']
```

```python
from sklearn.preprocessing import LabelEncoder


encoder = LabelEncoder()
encoded_feature = encoder.fit_transform(df[feature])

encoded_feature
```

```python
df['Result']= encoded_feature
df
```

```python
from sklearn.preprocessing import OneHotEncoder

feature1 = ['Gender']

encoder1 = OneHotEncoder()
encoded_feature1 = encoder1.fit_transform(df[feature1])
encoded_feature1.toarray()
```

```python
df[encoder1.get_feature_names_out()]= encoded_feature1.toarray()

df
```

```python
df.drop(columns= feature1,inplace= True)
df
```

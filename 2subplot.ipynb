```python
day=["sun","mon", "tue","wed","thu","fri","sat"]
share_price_ntc=[200,205,210,210,200,225,205]
share_price_ncell= [180,185,190,190,180,205,185]
share_price_nric = [100,165,170,130,160,185,165]
share_price_hrl = [140,145,150,110,140,165,145]
share_price_umhl = [120,125,120,130,120,195,125]
share_price_kbl = [100,105,120,110,100,145,105]
```

```python
from matplotlib import pyplot as plt


plt.figure(figsize =(10,5)) #10 inch in x axis 5 inch in y axis
plt.subplot(2,3,1)
plt.plot(day,share_price_ntc)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NTC")


plt.subplot(2,3,2) # Row,column, Data num
plt.plot(day,share_price_ncell)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NCELL")

plt.subplot(2,3,3)
plt.plot(day,share_price_nric)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NRIC")

plt.subplot(2,3,4)
plt.plot(day,share_price_hrl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("HRL")

plt.subplot(2,3,5)
plt.plot(day,share_price_umhl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("UMHL")

plt.subplot(2,3,6)
plt.plot(day,share_price_kbl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("KBL")


plt.tight_layout()  #Makes readable layout
plt.show()

```

```python
import numpy as np
import pandas as pd


train_data = pd.DataFrame({
    "Age": [20, 21, 22, 20, 23, 21, 22, np.nan, 19, 25],
    
    "Study_Hours": [
        5, 6, 4, np.nan, 7,
        5, 6, 4, 10, 5
    ],
    
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", np.nan, "Male", "Female", "Male"
    ],

    
})

print(train_data)

# numerical column => simpleimputer(strategy='mean'), meanmax()
# categorical column => simpleimputer(strategy='most_frequent'), onehotencoder()
```

```python
num_column = train_data.select_dtypes(include = ['int64', 'float64']).columns
cat_column = train_data.select_dtypes(include= ['str']).columns
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler


num_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'mean')),
    ('minmax', MinMaxScaler())
])
```

```python
from sklearn.preprocessing import OneHotEncoder


cat_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('onehot', OneHotEncoder())
])
```

```python
from sklearn.compose import ColumnTransformer


preprocessor = ColumnTransformer(
    [
        ('num', num_pipeline, num_column),
        ('cat', cat_pipeline , cat_column)
    ],
    verbose_feature_names_out = False
)
```

```python
value = preprocessor.fit_transform(train_data)
train_data[preprocessor.get_feature_names_out()] = value


train_data.drop(columns = ['Gender'], inplace = True)
train_data

```
